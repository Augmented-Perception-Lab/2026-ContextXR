# 01 · Explore MineXR++

ContextXR represents an app's functionality as **functional facets**: groups of capabilities that serve one user intent (paper §3.1). Within an app, facets form a **facet graph**. Navigational capabilities named `open_<facet>` link one facet to another, and each app has an *Open Section* entry facet.

**MineXR++** (§3.2) links each element that participants placed in their envisioned XR layouts to a facet and to its context of use: the user, the task, and the environment.

```python
from contextxr import load_minexrpp

ds = load_minexrpp()
cat = ds.catalog
print(f"{len(cat.apps)} apps, {len(cat)} facets, {len(cat.capability_names())} capabilities, "
      f"{len(ds)} scenarios, {len(ds.participants())} participants")
```

## A facet and its capabilities

Capabilities named `open_<facet>` are navigational. Each one is an edge of the facet graph.

```python
facet = cat.facet("recipes.view_recipe")
print(facet.name, "-", facet.description)
for capability in facet.capabilities:
    print("  ", capability, "(navigational)" if capability.startswith("open_") else "")
```

## The facet graph of one app (Fig. 2)

Manual navigation starts at the entry facet. The navigation cost of a facet (§3.4.1) is the number of facets on the shortest path from the entry facet to it, counting the entry facet.

```python
graph = cat.graph("recipes")
for edge in graph.edges:
    print(f"{edge.source:24} --{edge.capability}--> {edge.target}")
print()
for facet_id in graph.nodes:
    print(f"{facet_id:24} path from entry: {' > '.join(graph.path_from_entry(facet_id))}")
```

## Contexts and target facet sets

A context is the triple q = (u, t, e) (§3.3). The facets a participant placed for that context form the target set Y_q. Each facet carries the relevance tier the participant gave it.

```python
for context in ds.contexts_for("P04"):
    print(context)
```

```python
scenario = ds.scenarios["P04_LivingRoom_Relaxing"]
print(scenario.context)
for facet_id, tier in scenario.targets:
    print(f"  {tier:10} {facet_id}")
```
